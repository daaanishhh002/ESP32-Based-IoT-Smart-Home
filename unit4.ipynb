### Sampling and Quantization

```python
# sampling and quantization

t = np.linspace(0,10,11)
y = 0.85 ** t
plt.figure(figsize=(10,8))

# analog signal
plt.subplot(2,2,1)
plt.title('Analog Signal')
plt.xlabel('time')
plt.ylabel('amplitude')
plt.plot(t,y);

# sampling
plt.subplot(2,2,2)
plt.title('Sampling')
plt.xlabel('n')
plt.ylabel('amplitude')
plt.plot(t,y)
plt.stem(t,y);

# quantization
plt.subplot(2,2,3)
plt.title('Quantizing')
plt.xlabel('n')
plt.ylabel('range of quantizer')
for i in np.arange(0.1,1.1,0.1):
    plt.axhline(y = i, xmin = 0, xmax = 10, color = 'r')
plt.plot(t,y)
plt.stem(t,y);

# quantization
plt.subplot(2,2,4)
plt.title('Quantized Signal')
plt.xlabel('n')
plt.ylabel('range of quantizer')
for i in np.arange(0.1,1.1,0.1):
    plt.axhline(y = i, xmin = 0, xmax = 10, color = 'r')
plt.stem(t,np.round(y,1))

plt.tight_layout()
```

### Fourier Transform

```python
samp_freq = 100
samp_int = 1/samp_freq
plt.figure(figsize=(10,7))

t = np.arange(0,10,samp_int)
w1 = np.sin(2*np.pi*4*t)
w2 = np.sin(2*np.pi*7*t)
w0 = w1+w2

# 1st sine wave
plt.subplot(4,1,1)
plt.title('Sine wave with 4hz frequency')
plt.xlabel('time')
plt.ylabel('amplitude')
plt.plot(t,w1)

# 2nd sine wave
plt.subplot(4,1,2)
plt.title('Sine wave with 7hz frequency')
plt.xlabel('time')
plt.ylabel('amplitude')
plt.plot(t,w2)

# summed sine wave
plt.subplot(4,1,3)
plt.title('Sine wave with multiple frequency')
plt.xlabel('time')
plt.ylabel('amplitude')
plt.plot(t,w0)

# fourier transform of summed sine wave
ft = np.fft.fft(w0) / len(w0)
ft = ft[:len(w0) // 2]
frequencies = np.fft.fftfreq(len(w0), d=1/samp_freq)[:len(w0) // 2]

plt.subplot(4,1,4)
plt.title('Fourier Transform')
plt.xlabel('time')
plt.ylabel('amplitude')
plt.plot(frequencies,abs(ft))

plt.tight_layout()
```

### DFT and IDFT

**DFT:**
$$
X[k] = \sum_{n=0}^{N-1} x[n] \cdot e^{-j \frac{2\pi}{N} kn}
$$

```python
def DFT(x):

    N = len(x)
    n = np.arange(N)
    k = n.reshape((N, 1))
    e = np.exp(-2j * np.pi * k * n / N)
    
    X = np.dot(e, x)
    
    return X
```

**IDFT:**
$$
x[n] = \frac{1}{N} \sum_{k=0}^{N-1} X[k] \cdot e^{j \frac{2\pi}{N} kn}
$$

```python
def IDFT(X):

    N = len(X)
    n = np.arange(N)
    k = n.reshape((N, 1))
    e = np.exp(2j * np.pi * k * n / N)
    
    x = np.dot(e, X)
    x = x*(1/N)
    
    return x
```

#### DFT and IDFT for Discrete Values

```python
x = [1, 0, 1, 0, 1, 0, 1, 0]
X = DFT(x)
print("DFT values:",np.round(X,1))
```

```python
x = IDFT(X)
print(np.round(x,1))
```

#### Single Frequency Sine Wave

```python
# signal coonstruction

samp_freq = 50
samp_int = 1.0/samp_freq
freq = 1

t = np.arange(0,1,samp_int)
x = 3*np.sin(2*np.pi*freq*t)

plt.title('Original Signal')
plt.xlabel('time')
plt.ylabel('amplitude')
plt.plot(t,x);
```

```python
# dft of signal w/o fft

X = DFT(x)
N = len(X)
n = np.arange(N)

T = N/samp_freq
freq = n/T 

plt.title('DFT')
plt.xlabel('freq')
plt.ylabel('|X(freq)|')
plt.stem(freq,abs(X),markerfmt="");
```

```python
# signal reconstruction by idft w/o fft

samp_freq = 50
samp_int = 1.0/samp_freq

x = IDFT(X)
t = np.arange(0,1,samp_int)

plt.title('IDFT')
plt.xlabel('freq')
plt.ylabel('|x(freq)|')
plt.plot(t,x);
```

#### Multiple Frequency Sine Wave

```python
# signal coonstruction


samp_freq = 100
samp_int = 1.0/samp_freq
t = np.arange(0,1,samp_int)

freq = 1.0
x = 3*np.sin(2*np.pi*freq*t)

freq = 4.0
x += np.sin(2*np.pi*freq*t)

freq = 7.0 
x += 0.5* np.sin(2*np.pi*freq*t)

plt.title('Multi Frequency Signal')
plt.xlabel('time')
plt.ylabel('amplitude')
plt.plot(t,x);
```

```python
# dft of signal w/o fft

X = DFT(x)
N = len(X)
n = np.arange(N)
T = N/samp_freq
freq = n/T

plt.xlabel('freq')
plt.ylabel('|X(freq)|')
plt.stem(freq,abs(X),markerfmt="");
```

```python
# signal reconstruction by idft w/o fft

x=IDFT(X)
samp_freq = 100
samp_int = 1.0/samp_freq
t = np.arange(0,1,samp_int)

 
plt.xlabel('freq')
plt.ylabel('|X(freq)|')
plt.plot(t,x);
```

### FFT and IFFT

```python
# dft using fft

x = [1, 0, 1, 0, 1, 0, 1, 0]
fft_data = np.fft.fft(x)
mag = np.abs(fft_data)
plt.figure(figsize=(8,4))

# original signal
plt.subplot(1, 3, 1)
plt.xlabel('n')
plt.ylabel('amplitude')
plt.stem(x);

# dft
plt.subplot(1, 3, 2)
plt.xlabel('n')
plt.ylabel('mag')
plt.plot(mag);

# reconstructiion using ifft
ifft_data = np.fft.ifft(fft_data)

plt.subplot(1, 3, 3)
plt.xlabel('n')
plt.ylabel('amplitude')
plt.stem(ifft_data);

plt.tight_layout();
```

### LT and ILT

#### Using `sympy`

```python
t, s = smp.symbols('t, s')
```

```python
def L(f):
    return smp.laplace_transform(f, t, s, noconds=True)
```

```python
def invL(F):
    return smp.inverse_laplace_transform(F, s, t)
```

**Laplace Transform of:**
$$e^{-at}$$

```python
x, y = symbols('a t')
exp = smp.exp(-x*y)
exp
```

```python
pxe = L(exp)
pxe
```

```python
invL(pxe)
```

### Z Transform and Inverse Z Transform

#### Signals

**Z Transform of:**
$$𝛿[𝑛]+2𝛿[𝑛−2]$$

```python
from lcapy import delta, n, z

# original

x = delta(n) + 2 * delta(n - 2)
x
```

```python
# z transform

ans = x(z)
ans
```

```python
# inverse z transform

ans(n)
```

**Z Transform of:**
$$2^{−𝑛}𝑢[𝑛]$$

```python
from lcapy import u

# original

x = (0.5**n)*u(n)
x
```

```python
# z transform

ans = x(z)
ans
```

```python
# inverse z transform

ans(n)
```
