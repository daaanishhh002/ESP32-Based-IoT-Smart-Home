### DFT and IDFT

**DFT:**
$$
X[k] = \sum_{n=0}^{N-1} x[n] \cdot e^{-j \frac{2\pi}{N} kn}
$$

```python
def DFT(x):

    N = len(x)
    n = np.arange(N)
    k = n.reshape((N, 1))
    e = np.exp(-2j * np.pi * k * n / N)
    
    X = np.dot(e, x)
    
    return X
```

**IDFT:**
$$
x[n] = \frac{1}{N} \sum_{k=0}^{N-1} X[k] \cdot e^{j \frac{2\pi}{N} kn}
$$

```python
def IDFT(X):

    N = len(X)
    n = np.arange(N)
    k = n.reshape((N, 1))
    e = np.exp(2j * np.pi * k * n / N)
    
    x = np.dot(e, X)
    x = x*(1/N)
    
    return x
```

#### DFT and IDFT for Discrete Values

```python
x = [1, 0, 1, 0, 1, 0, 1, 0]
X = DFT(x)
print("DFT values:",np.round(X,1))
```

```python
x = IDFT(X)
print(np.round(x,1))
```

### FIR Filter

```python
def mfreqz(b,a=1):
    w,h = signal.freqz(b,a)
    h_dB = 20 * np.log10 (abs(h))
    
    plt.subplot(1,2,1)
    plt.plot(w/max(w),h_dB)
    plt.ylabel('Magnitude (db)')
    plt.xlabel('Normalized Frequency')
    plt.title('Frequency response')
    
    plt.subplot(1,2,2)
    h_Phase = np.unwrap(np.arctan2(np.imag(h),np.real(h)))
    plt.plot(w/max(w),h_Phase)
    plt.ylabel('Phase (radians)')
    plt.xlabel('Normalized Frequency')
    plt.title('Phase response')
    plt.tight_layout()
```

Write a python code to design FIR filters (LP/HP/BP/BS) using windowing method for the following specifications cut-off frequency = 0.2 and order= 9.

```python
from scipy import signal

freq1 = 0.2
freq2 = 0.5
order = 9

sig1 = 2*np.pi*50
sig2 = 2*np.pi*200
```

## Part D

- Cleaning Up Data Noise with Fourier Transform

```python
# create signal with two frequencies

data_step = 0.001
t = np.arange(0,1,data_step)
f_clean = np.sin(2*np.pi*50*t) + np.sin(2*np.pi*120*t)
f_noise = f_clean + 2.5*np.random.randn(len(t))
plt.plot(t,f_noise,color='c',linewidth=1.5,label='Noisy')
plt.plot(t,f_clean,color='k',linewidth=2,label='Clean')
plt.legend()
plt.show()
```

```python
# time domain to frequency domain
n = len(t)
yf = sp.fft.rfft(f_noise)
xf = sp.fft.rfftfreq(n,data_step)
plt.plot(xf,np.abs(yf))
plt.show();
```

```python
# remove noise frequencies
yf_abs = np.abs(yf) 
indices = yf_abs>300   # filter out those value under 300
yf_clean = indices * yf # noise frequency will be set to 0
plt.plot(xf,np.abs(yf_clean))
plt.show()
```

```python
# inverse transform to plot denoised wave
from scipy.fft import irfft
new_f_clean = irfft(yf_clean)
plt.plot(t,new_f_clean)
plt.ylim(-6,8)
plt.show()
```
